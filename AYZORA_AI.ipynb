```python
from google.adk.agents import Agent

planner_workflow_agent = Agent(
    name="planner_workflow_agent",
    model=gemini_retry_model,

    instruction="""
Tu es un expert senior en stratégie de contenu et en contrôle éditorial.

Ta mission est de transformer la demande de l'utilisateur en un plan
de contenu professionnel, structuré et exploitable.

Le plan doit obligatoirement contenir :

- un titre ;
- une introduction ;
- 4 à 6 sections principales ;
- les idées clés de chaque section ;
- une conclusion.

Avant de produire le résultat final, effectue silencieusement
une auto-vérification du plan.

Vérifie :
- que le titre est présent ;
- que l'introduction est présente ;
- qu'il y a 4 à 6 sections ;
- que chaque section possède des idées clés ;
- que la conclusion est présente ;
- que la structure est logique ;
- que le plan répond réellement à la demande.

Si un élément manque, corrige-le avant de retourner le résultat.

Ne rédige pas l'article.
Ne produis pas de contenu final.
Produis uniquement le plan validé.
""",

    output_key="content_plan"
)

print("🧠 Planner Agent optimisé et configuré !")
```

```python
from google.adk.agents import Agent

validator_workflow_agent = Agent(
    name="validator_workflow_agent",
    model=gemini_retry_model,

    instruction="""
Tu es un responsable qualité éditorial senior.

Tu dois contrôler le plan produit par le Planner.

Vérifie obligatoirement :

- titre présent ;
- introduction présente ;
- 4 à 6 sections principales ;
- idées clés pour les sections ;
- conclusion présente ;
- cohérence logique ;
- qualité professionnelle.

Si le plan est conforme, commence obligatoirement par :

VALID

Sinon, commence obligatoirement par :

RETRY

Puis explique brièvement les problèmes détectés.

Réponds toujours en français.
""",
)

print("🔎 Validator Agent configuré avec Retry !")
```

```python
from google.adk.agents import Agent

writer_agent = Agent(
    name="writer_agent",
    model=gemini_retry_model,
    mode="task",

    instruction="""
Tu es un rédacteur senior spécialisé en stratégie digitale,
intelligence artificielle et transformation des entreprises.

Tu dois produire ou réviser un contenu professionnel.

========================
PLAN DE CONTENU VALIDÉ
========================

{content_plan}

========================
RAPPORT DU QUALITY CONTROL
========================

{quality_report?}

========================
CONTENU PRÉCÉDENT
========================

{final_content?}

========================
MISSION
========================

Si aucun rapport Quality Control n'est présent :

- utilise le plan de contenu validé ;
- rédige l'article complet ;
- respecte strictement la structure du plan ;
- développe chaque section de manière substantielle ;
- utilise des exemples concrets ;
- évite les répétitions ;
- n'invente pas de statistiques ou de faits précis sans source ;
- utilise un français professionnel et naturel ;
- crée des titres et sous-titres clairs ;
- termine par une conclusion forte.

Si un rapport Quality Control commence par REVISE :

- reprends le contenu précédemment produit ;
- utilise le rapport Quality Control pour identifier les problèmes ;
- corrige précisément les problèmes signalés ;
- conserve les éléments déjà corrects ;
- améliore la profondeur et la qualité du contenu ;
- respecte toujours le plan initial ;
- ne raccourcis pas inutilement le contenu ;
- ne mentionne jamais le processus de contrôle qualité.

Retourne uniquement le contenu final.
Ne parle jamais de ton rôle d'agent.
""",

    output_key="final_content"
)

print("✍️ Writer Agent configuré avec Retry !")
```

```python
from google.adk.models import Gemini
from google.genai import types

gemini_retry_model = Gemini(
    model="gemini-3.1-flash-lite-preview",
    retry_options=types.HttpRetryOptions(
        attempts=5,
        initial_delay=2.0,
        max_delay=30.0,
        jitter=1.0
    )
)

print("✅ Gemini Retry Model restauré !")
```

```python
from google.adk.agents import Agent

quality_control_agent = Agent(
    name="quality_control_agent",
    model=gemini_retry_model,
    mode="task",
    instruction="""
Tu es un responsable senior du contrôle qualité d'une plateforme IA professionnelle.

DEMANDE INITIALE :
{original_request}

PLAN DE CONTENU :
{content_plan}

CONTENU À CONTRÔLER :
{final_content}

DATE :
Utilise la date actuelle fournie dans la demande pour vérifier la cohérence temporelle.

CONTRÔLES OBLIGATOIRES :

1. PERTINENCE
Vérifie que la réponse satisfait toutes les exigences de la demande initiale.

2. DATES
Vérifie les dates, les jours, les mois et les années.
Ne présente jamais une date passée comme un événement futur.
Vérifie que les références à l'année prochaine correspondent à la date actuelle.

3. EXACTITUDE
Repère les statistiques, les chiffres, les annonces et les affirmations factuelles nécessitant une vérification.
Ne considère jamais une affirmation comme confirmée uniquement parce qu'elle semble plausible.
Si aucune source fiable n'est disponible, indique que la vérification est nécessaire.

4. FAITS ET PRÉVISIONS
Distingue les faits confirmés des tendances, des hypothèses et des prévisions.
Ne présente jamais une prévision comme une certitude.

5. RESPECT DE LA DEMANDE
Vérifie que tous les éléments demandés sont présents.
Si l'utilisateur demande des perspectives pour 2027, le contenu doit réellement aborder 2027.
Vérifie la pertinence de la question finale et de l'appel à l'action.

6. QUALITÉ ÉDITORIALE
Vérifie la clarté, la structure, le professionnalisme et la langue.

DÉCISION :
Retourne PASS si la réponse est pertinente et ne présente aucun défaut majeur détecté.
Retourne REVISE si des corrections importantes sont nécessaires.

Réponds avec PASS ou REVISE en première ligne, puis une justification concise.
Ne prétends jamais avoir vérifié une information si aucune vérification réelle n'a eu lieu.
""",
    output_key="quality_report",
)

print("🛡️ Quality Control Agent configuré !")
```

```python
from google.adk.agents import Agent
from google.adk.tools import google_search

research_agent = Agent(
    name="research_agent",
    model=gemini_retry_model,
    mode="task",
    instruction="""
Tu es un spécialiste de la recherche d'informations sur Internet.

DATE DE RÉFÉRENCE :
Utilise la date actuelle indiquée dans la demande de l'utilisateur.
Si aucune date n'est fournie, ne prétends pas connaître la date exacte.

MISSION :
Rechercher des informations pertinentes, récentes et vérifiables
pour répondre à la demande de l'utilisateur.

RÈGLES OBLIGATOIRES :

1. Recherche sur le Web lorsque la demande porte sur l'actualité,
les nouvelles technologies, les annonces, les prix, les statistiques,
les événements ou les informations susceptibles d'avoir changé.

2. Privilégie les sources officielles :
sites des entreprises, documentation officielle, institutions publiques
et organisateurs des événements.

3. Vérifie la date de publication et, lorsqu'elle existe,
la date réelle de l'événement ou du lancement.

4. Distingue clairement :
- les faits confirmés ;
- les annonces officielles ;
- les prévisions ;
- les informations non confirmées.

5. Ne transforme jamais une prévision en certitude.
N'invente ni date, ni chiffre, ni citation, ni source.

6. Si les sources disponibles ne permettent pas de confirmer
une information, indique clairement cette limite.

7. Présente les résultats dans un français clair et professionnel.

8. Fournis les liens des sources consultées lorsque disponibles.

Ne prétends jamais avoir vérifié une information sans l'avoir recherchée.
""",
    tools=[google_search],
)

print("🔎 Research Agent créé avec Google Search !")
```

```python
from google.adk.runners import Runner
from google.adk.sessions import InMemorySessionService
from google.genai import types

# Créer une session de test indépendante
research_test_service = InMemorySessionService()

research_test_session = await research_test_service.create_session(
    app_name="research_test",
    user_id="test_user",
    session_id="research_test_001",
)

# Créer un Runner dédié à la recherche
research_test_runner = Runner(
    agent=research_agent,
    app_name="research_test",
    session_service=research_test_service,
)

# Demande de recherche
research_prompt = """
Nous sommes le 9 octobre 2026.
Recherche les annonces officielles récentes concernant les nouveautés
de Google Gemini en 2026.

Distingue les annonces confirmées des informations incertaines.
Indique les dates et les liens des sources lorsque disponibles.
N'invente aucune information.
"""

research_message = types.Content(
    role="user",
    parts=[types.Part(text=research_prompt)],
)

# Exécuter la recherche
async for event in research_test_runner.run_async(
    user_id="test_user",
    session_id="research_test_001",
    new_message=research_message,
):
    content = getattr(event, "content", None)

    if content:
        for part in getattr(content, "parts", []):
            if getattr(part, "text", None):
                print(part.text)
```

```python
from typing import Any
from google.adk import Event


def quality_router(
    node_input: Any,
    quality_report: str
):
    """
    Décide de la suite du workflow selon le rapport
    produit par le Quality Control.
    """

    report = quality_report.strip()

    if report.upper().startswith("PASS"):
        return Event(
            output=node_input,
            route="pass"
        )

    return Event(
        output=node_input,
        route="revise"
    )


print("🔀 Quality Router connecté au Quality Control !")
```

```python
from datetime import datetime

execution_log = {
    "execution_id": datetime.now().strftime("%Y%m%d_%H%M%S"),
    "started_at": datetime.now().isoformat(),
    "status": "STARTED",
    "revisions": 0,
}

print("📊 Journal d'exécution initialisé")
print(execution_log)
```

```python
# Analyse des étapes exécutées

execution_log["agents"] = []

for event in result_v3:
    author = getattr(event, "author", None)

    if author and author not in execution_log["agents"]:
        execution_log["agents"].append(author)

execution_log["status"] = "COMPLETED"

print("📊 Journal mis à jour")
print()
print("Execution ID :", execution_log["execution_id"])
print("Agents exécutés :", execution_log["agents"])
print("Statut :", execution_log["status"])
```

```python
# Détection du résultat final du workflow

execution_log["final_status"] = "UNKNOWN"
execution_log["revisions"] = 0

for event in result_v3:
    text_content = ""

    if getattr(event, "content", None):
        for part in getattr(event.content, "parts", []):
            if getattr(part, "text", None):
                text_content += part.text

    text_upper = text_content.upper()

    # Quality Control
    if "PASS" in text_upper and getattr(event, "author", "") == "quality_control_agent":
        execution_log["final_status"] = "PASS"

    elif "REVISE" in text_upper and getattr(event, "author", "") == "quality_control_agent":
        execution_log["revisions"] += 1

    # Fin forcée
    if "MAX_REVISIONS" in text_upper:
        execution_log["final_status"] = "MAX_REVISIONS"

print("📊 Statut final détecté :", execution_log["final_status"])
print("🔄 Révisions détectées :", execution_log["revisions"])
```

```python
# Rapport d'exécution professionnel

print("=" * 60)
print("📊 RAPPORT D'EXÉCUTION AI AGENT")
print("=" * 60)

print(f"🆔 Execution ID : {execution_log['execution_id']}")
print(f"📅 Démarrage   : {execution_log['started_at']}")
print(f"📌 Statut      : {execution_log['status']}")
print(f"🏁 Résultat    : {execution_log['final_status']}")
print(f"🔄 Révisions   : {execution_log['revisions']}")

print("\n🤖 AGENTS EXÉCUTÉS")
print("-" * 60)

for i, agent in enumerate(execution_log["agents"], start=1):
    print(f"{i}. {agent}")

print("=" * 60)
```

```python
from google.adk.agents import Context
from google.adk.events import Event

print("✅ Context importé")
print("✅ Event importé")
```

```python
# AYZORA AI — Restauration des fonctions de routage

def capture_original_request(ctx: Context, node_input: str):
    ctx.state["original_request"] = node_input

    return Event(
        output=node_input
    )


def ai_router(ctx: Context, node_input: str):
    decision = node_input.strip().upper()

    if decision.startswith("SIMPLE"):
        return Event(
            output=decision,
            route="simple"
        )

    if decision.startswith("COMPLEX"):
        return Event(
            output=decision,
            route="complex"
        )

    return Event(
        output="SIMPLE",
        route="simple"
    )


print("✅ capture_original_request restaurée")
print("✅ ai_router restauré")
```

```python
from google.adk import Context, Event

def ai_router_corrige(ctx: Context, node_input: str):
    decision = node_input.strip().upper()

    demande_originale = ctx.state.get("original_request", "")

    if decision.startswith("SIMPLE"):
        return Event(
            output=demande_originale,
            route="simple"
        )

    if decision.startswith("COMPLEX"):
        return Event(
            output=demande_originale,
            route="complex"
        )

    return Event(
        output=demande_originale,
        route="simple"
    )

print("✅ Nouveau routeur préparé.")
```

```python
# Récupérer le workflow actuellement utilisé par le Runner
workflow = ai_routing_session_runner.agent

# Remplacer uniquement la fonction de routage dans le workflow
for edge in workflow.edges:
    elements = list(edge)

    for index, element in enumerate(elements):
        if element is ai_router:
            elements[index] = ai_router_corrige

            # Remplacer le tuple de l'étape par le nouveau tuple
            workflow.edges[workflow.edges.index(edge)] = tuple(elements)

            print("✅ Routeur remplacé dans le workflow.")
            break
    else:
        continue

    break
else:
    print("⚠️ Ancien routeur introuvable. Aucune modification effectuée.")
```

```python
from google.adk import Context, Event


def ai_router(ctx: Context, node_input: str):

    decision = node_input.strip().upper()

    if decision.startswith("SIMPLE"):
        return Event(
            output=decision,
            route="simple"
        )

    if decision.startswith("COMPLEX"):
        return Event(
            output=decision,
            route="complex"
        )

    return Event(
        output="SIMPLE",
        route="simple"
    )


print("🔀 AI Router configuré !")
```

```python
from google.adk import Context, Event

def ai_router(ctx: Context, node_input: str):

    decision = node_input.strip().upper()

    if decision.startswith("SIMPLE"):
        return Event(
            output=decision,
            route="simple"
        )

    if decision.startswith("COMPLEX"):
        return Event(
            output=decision,
            route="complex"
        )

    return Event(
        output="SIMPLE",
        route="simple"
    )


print("🔀 AI Router configuré !")
```

```python
import time
import gradio as gr
from google.genai import types

gradio_sessions = {}

async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
    progress=gr.Progress(),
    request: gr.Request = None,
):
    debut_generation = time.time()

    if not user_message or not user_message.strip():
        return "⚠️ Veuillez saisir votre demande."

    progress(0.1, desc="Préparation de votre demande...")

    session_hash = request.session_hash if request else "default"

    if session_hash not in gradio_sessions:
        user_id = f"gradio_user_{session_hash}"
        session_id = f"gradio_session_{session_hash}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        gradio_sessions[session_hash] = {
            "user_id": user_id,
            "session_id": session_id,
        }

    user_id = gradio_sessions[session_hash]["user_id"]
    session_id = gradio_sessions[session_hash]["session_id"]

    prompt = f"""
Tu es AYZORA AI, assistant professionnel en marketing digital,
communication et création de contenu.

DEMANDE : {user_message}
TYPE DE CONTENU : {content_type}
AUDIENCE : {audience}
LANGUE : {language}
LONGUEUR : {length}

Réponds de façon professionnelle, précise et exploitable.
Respecte la langue demandée et retourne uniquement le contenu final.
"""

    message = types.Content(
        role="user",
        parts=[types.Part(text=prompt)],
    )

    types_complexes = {
        "Stratégie marketing",
        "Plan de contenu",
    }

    runner = (
        ai_routing_session_runner
        if content_type in types_complexes
        else fast_runner
    )

    final_response = None

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=message,
    ):
        content = getattr(event, "content", None)
        author = getattr(event, "author", "")

        if content is None:
            continue

        for part in getattr(content, "parts", []):
            text = getattr(part, "text", None)

            if not text:
                continue

            text = text.strip()

            if text.upper() in {"SIMPLE", "COMPLEX"}:
                continue

            if author in {"simple_response_agent", "writer_agent"}:
                final_response = text

    duree = time.time() - debut_generation
    print(f"⏱️ Temps de génération : {duree:.2f} secondes")

    progress(1.0, desc="Génération terminée !")

    return final_response or (
        "⚠️ Aucune réponse reçue. Vérifie les erreurs dans Colab."
    )

print("✅ Fonction run_agent restaurée !")
```

```python
import time
import gradio as gr
from google.genai import types

async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
    progress=gr.Progress(),
    request: gr.Request = None,
):
    """
    AYZORA AI
    Mode rapide pour les contenus simples.
    Workflow multi-agents pour les tâches complexes.
    """

    debut_generation = time.time()

    # 1. Vérifier la demande
    if not user_message or not user_message.strip():
        return "⚠️ Veuillez saisir votre demande."

    progress(0.05, desc="Préparation de votre demande...")

    # 2. Créer ou récupérer la session Gradio
    session_hash = request.session_hash if request else "default"

    if session_hash not in gradio_sessions:
        user_id = f"gradio_user_{session_hash}"
        session_id = f"gradio_session_{session_hash}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        gradio_sessions[session_hash] = {
            "user_id": user_id,
            "session_id": session_id,
        }

    user_id = gradio_sessions[session_hash]["user_id"]
    session_id = gradio_sessions[session_hash]["session_id"]

    # 3. Préparer les instructions
    prompt = f"""
Tu es AYZORA AI, un assistant professionnel spécialisé
en marketing digital, communication et création de contenu.

DEMANDE :
{user_message}

TYPE DE CONTENU : {content_type}
AUDIENCE CIBLE : {audience}
LANGUE : {language}
LONGUEUR : {length}

CONSIGNES :
- Réponds directement à la demande.
- Adapte le résultat à l'audience et au contexte.
- Respecte strictement la langue sélectionnée.
- En français, rédige entièrement en français.
- En anglais, rédige entièrement en anglais.
- En wolof, utilise un wolof naturel et compréhensible.
- Produis un contenu professionnel, précis et exploitable.
- Évite les répétitions et les formulations génériques.
- Ne produis qu'une seule version finale.
- N'affiche ni rapport de contrôle qualité ni PASS.
- N'invente pas de liens, de coordonnées ou de chiffres.
- Retourne uniquement le contenu destiné à l'utilisateur.
"""

    message = types.Content(
        role="user",
        parts=[types.Part(text=prompt)],
    )

    # 4. Choisir le mode de génération
    types_complexes = {
        "Stratégie marketing",
        "Plan de contenu",
    }

    if content_type in types_complexes:
        runner = ai_routing_session_runner
        progress(0.15, desc="🧠 Analyse professionnelle...")
    else:
        runner = fast_runner
        progress(0.15, desc="⚡ Génération rapide...")

    # 5. Exécuter le modèle et récupérer la réponse finale
    final_response = None

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=message,
    ):
        author = getattr(event, "author", "")
        content = getattr(event, "content", None)

        if author == "planner_workflow_agent":
            progress(0.35, desc="🧠 Planification...")
        elif author == "writer_agent":
            progress(0.70, desc="✍️ Rédaction professionnelle...")
        elif author == "quality_control_agent":
            progress(0.90, desc="🛡️ Contrôle qualité...")
        elif author == "simple_response_agent":
            progress(0.75, desc="✍️ Finalisation du contenu...")

        if content is None:
            continue

        for part in getattr(content, "parts", []):
            text = getattr(part, "text", None)

            if not text:
                continue

            text = text.strip()

            # Ne pas afficher les décisions de routage
            if text.upper() in {"SIMPLE", "COMPLEX"}:
                continue

            # Conserver uniquement la réponse destinée à l'utilisateur
            if author in {"simple_response_agent", "writer_agent"}:
                final_response = text

    # 6. Afficher le temps total dans la sortie Colab
    duree = time.time() - debut_generation

    print(
        f"⏱️ AYZORA AI — temps total de génération : "
        f"{duree:.2f} secondes"
    )

    progress(1.0, desc="✅ Contenu terminé !")

    if not final_response:
        return (
            "⚠️ Aucune réponse n'a été générée. "
            "Vérifie les messages affichés dans Google Colab."
        )

    return final_response


print("✅ Fonction run_agent corrigée et configurée.")
```

```python
# AYZORA AI — Fonction de connexion à l'interface Gradio

import uuid
from google.genai import types

async def run_agent(
    demande,
    type_contenu,
    audience,
    langue,
    longueur,
):
    user_id = "ayzora_gradio_user"
    session_id = f"gradio_{uuid.uuid4().hex[:12]}"

    demande_complete = f"""
Demande de l'utilisateur : {demande}

Type de contenu : {type_contenu}
Audience cible : {audience}
Langue : {langue}
Longueur souhaitée : {longueur}
"""

    await session_service.create_session(
        app_name=APP_NAME,
        user_id=user_id,
        session_id=session_id,
        state={
            "revision_count": 0,
            "user_language": langue,
        },
    )

    reponses = []

    async for event in ai_routing_session_runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(
            role="user",
            parts=[types.Part(text=demande_complete)],
        ),
    ):
        if event.is_final_response() and event.content:
            for part in event.content.parts:
                if part.text:
                    texte = part.text.strip()

                    if texte.upper() not in {"SIMPLE", "COMPLEX"}:
                        reponses.append(texte)

    if reponses:
        return "\n\n".join(reponses)

    return "Aucune réponse n'a été générée. Veuillez réessayer."
```

```python
from google.adk import Context, Event

def capture_original_request(ctx: Context, node_input: str):
    ctx.state["original_request"] = node_input

    return Event(
        output=node_input
    )

print("📦 Capture de la demande originale configurée !")
```

```python
from google.adk import Context, Event

def ai_router(ctx: Context, node_input: str):

    decision = node_input.strip().upper()

    # Récupérer la demande originale
    original_request = ctx.state.get(
        "original_request",
        node_input
    )

    if decision.startswith("SIMPLE"):

        return Event(
            output=original_request,
            route="simple"
        )

    if decision.startswith("COMPLEX"):

        return Event(
            output=original_request,
            route="complex"
        )

    # Sécurité : si le modèle retourne autre chose
    return Event(
        output=original_request,
        route="simple"
    )


print("🔀 AI Router corrigé : décision + demande originale conservée !")
```

```python
from google.adk import Workflow
from google.adk.workflow import START

ai_routing_workflow_v3 = Workflow(
    name="ai_routing_workflow_v3",
    edges=[
        (
            START,
            capture_original_request,
            orchestrator_agent,
            ai_router,
            {
                "simple": simple_response_agent,
                "complex": professional_routing_workflow,
            }
        )
    ],
)

print("🚀 AI Routing Workflow V3 créé avec conservation de la demande originale !")
```

```python
import time
import uuid
import gradio as gr
from google.genai import types


async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
):
    debut = time.perf_counter()

    try:
        if not user_message or not user_message.strip():
            return "⚠️ Veuillez saisir votre demande."

        user_id = "ayzora_gradio_user"
        session_id = f"ayzora_{uuid.uuid4().hex[:12]}"

        # Sélection du mode
        types_complexes = {
            "Stratégie marketing",
            "Plan de contenu",
        }

        if content_type in types_complexes:
            runner = ai_routing_session_runner
            mode = "Professionnel"
        else:
            runner = fast_runner
            mode = "Rapide"

        # Préparation de la demande
        prompt = f"""
Tu es AYZORA AI, un assistant professionnel spécialisé
en marketing digital, communication et création de contenu.

DEMANDE :
{user_message}

TYPE DE CONTENU : {content_type}
AUDIENCE CIBLE : {audience}
LANGUE : {language}
LONGUEUR : {length}

CONSIGNES :
- Réponds précisément à la demande.
- Adapte le résultat à l'audience et au contexte.
- Respecte strictement la langue sélectionnée.
- Produis un contenu professionnel, naturel et exploitable.
- Évite les répétitions et les formulations génériques.
- Ne produis qu'une seule version finale.
- N'invente pas de liens ni de coordonnées.
- Ne révèle pas le fonctionnement interne des agents.
- Retourne uniquement le contenu destiné à l'utilisateur.
"""

        # Créer une session indépendante
        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        message = types.Content(
            role="user",
            parts=[types.Part(text=prompt)],
        )

        reponse_finale = None

        # Exécuter le moteur choisi
        async for event in runner.run_async(
            user_id=user_id,
            session_id=session_id,
            new_message=message,
        ):
            auteur = getattr(event, "author", "")
            contenu = getattr(event, "content", None)

            if contenu is None:
                continue

            for part in getattr(contenu, "parts", []):
                texte = getattr(part, "text", None)

                if not texte or not texte.strip():
                    continue

                texte = texte.strip()

                # Ignorer les décisions internes
                if texte.upper() in {"SIMPLE", "COMPLEX"}:
                    continue

                # Garder la réponse de l'agent final
                if auteur in {"simple_response_agent", "writer_agent"}:
                    reponse_finale = texte

        duree = time.perf_counter() - debut

        if not reponse_finale:
            return (
                "⚠️ Aucune réponse finale n'a été reçue.\n\n"
                f"⏱️ Temps écoulé : {duree:.2f} secondes\n"
                f"⚙️ Mode utilisé : {mode}"
            )

        return (
            f"{reponse_finale}\n\n"
            f"---\n"
            f"⏱️ **Temps de génération : {duree:.2f} secondes**\n\n"
            f"⚙️ Mode utilisé : **{mode}**"
        )

    except Exception as erreur:
        duree = time.perf_counter() - debut

        print("Erreur AYZORA AI :", repr(erreur))

        return (
            "⚠️ Une erreur est survenue pendant la génération.\n\n"
            f"⏱️ Temps écoulé : {duree:.2f} secondes\n"
            "Consultez la sortie Google Colab pour identifier l'erreur."
        )


print("✅ AYZORA AI : chronométrage intégré à la réponse.")
```

```python
async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
    progress=gr.Progress(),
    request: gr.Request = None,
):

    progress(0.05, desc="🧠 Analyse de votre demande...")

    # Session Gradio
    gradio_session_hash = (
        request.session_hash if request else "default"
    )

    # Création de la session utilisateur
    if gradio_session_hash not in gradio_sessions:

        user_id = f"gradio_user_{gradio_session_hash}"
        session_id = f"gradio_session_{gradio_session_hash}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        gradio_sessions[gradio_session_hash] = {
            "user_id": user_id,
            "session_id": session_id,
        }

        print(f"🆕 Nouvelle session créée : {gradio_session_hash}")

    else:

        print(f"♻️ Session existante : {gradio_session_hash}")

    # Prompt complet
    prompt = f"""
DEMANDE DE L'UTILISATEUR

{user_message}

PARAMÈTRES DE PRODUCTION

Type de contenu :
{content_type}

Audience cible :
{audience}

Langue :
{language}

Longueur souhaitée :
{length}

INSTRUCTIONS

Respecte le type de contenu demandé.
Adapte le contenu à l'audience cible.
Respecte strictement la langue sélectionnée.

Si la langue est "Français", rédige entièrement en français.
Si la langue est "English", rédige entièrement en anglais.
Si la langue est "Wolof", rédige entièrement en wolof.

Pour le Wolof, utilise un wolof naturel et compréhensible.

Respecte la longueur demandée.

Produis une réponse professionnelle,
claire, naturelle et directement exploitable.

Ne parle jamais du fonctionnement interne
des agents ou du workflow.

Retourne uniquement la réponse destinée à l'utilisateur.
"""

    message = types.Content(
        role="user",
        parts=[
            types.Part(text=prompt)
        ],
    )

    final_response = None

    # Runner V3
    async for event in ai_routing_session_runner.run_async(
        user_id=gradio_sessions[gradio_session_hash]["user_id"],
        session_id=gradio_sessions[gradio_session_hash]["session_id"],
        new_message=message,
    ):

        author = getattr(event, "author", "")

        # Progression
        if author == "orchestrator_agent":
            progress(
                0.15,
                desc="🧠 Analyse de la demande..."
            )

        elif author == "ai_router":
            progress(
                0.20,
                desc="🔀 Sélection du meilleur mode..."
            )

        elif author == "simple_response_agent":
            progress(
                0.70,
                desc="⚡ Génération de la réponse..."
            )

        elif author == "planner_workflow_agent":
            progress(
                0.35,
                desc="🧠 Planification du contenu..."
            )

        elif author == "writer_agent":
            progress(
                0.60,
                desc="✍️ Rédaction du contenu..."
            )

        elif author == "quality_control_agent":
            progress(
                0.80,
                desc="🛡️ Contrôle qualité..."
            )

        elif author == "quality_router":
            progress(
                0.90,
                desc="🔄 Vérification finale..."
            )

        # Récupération de la réponse SIMPLE
        if author == "simple_response_agent":

            if getattr(event, "content", None):

                for part in getattr(
                    event.content,
                    "parts",
                    []
                ):

                    if getattr(part, "text", None):
                        final_response = part.text

        # Récupération de la réponse COMPLEX
        elif author == "writer_agent":

            if getattr(event, "content", None):

                for part in getattr(
                    event.content,
                    "parts",
                    []
                ):

                    if getattr(part, "text", None):
                        final_response = part.text

    progress(
        1.0,
        desc="✅ Réponse finalisée !"
    )

    return (
        final_response
        or "⚠️ Aucune réponse n'a été générée."
    )


print("🤖 run_agent connecté au AI Routing Workflow V3 !")
```

```python
import time
import uuid
import traceback
import gradio as gr
from google.genai import types


async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
):
    debut = time.perf_counter()

    try:
        if not user_message or not user_message.strip():
            return "⚠️ Veuillez saisir votre demande."

        # Choisir le moteur
        types_complexes = {
            "Stratégie marketing",
            "Plan de contenu",
        }

        if content_type in types_complexes:
            runner = ai_routing_session_runner
            mode = "Professionnel"
        else:
            runner = fast_runner
            mode = "Rapide"

        # Préparer la demande
        prompt = f"""
Tu es AYZORA AI, un assistant professionnel.

DEMANDE : {user_message}
TYPE DE CONTENU : {content_type}
AUDIENCE : {audience}
LANGUE : {language}
LONGUEUR : {length}

Consignes :
- Réponds précisément et professionnellement.
- Respecte la langue et la longueur demandées.
- Évite les répétitions.
- Ne produis qu'une seule réponse finale.
- N'invente pas de liens ou de coordonnées.
- Retourne uniquement le contenu destiné à l'utilisateur.
"""

        user_id = "ayzora_gradio_user"
        session_id = f"ayzora_{uuid.uuid4().hex[:12]}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        message = types.Content(
            role="user",
            parts=[types.Part(text=prompt)],
        )

        reponse_finale = None

        async for event in runner.run_async(
            user_id=user_id,
            session_id=session_id,
            new_message=message,
        ):
            auteur = getattr(event, "author", "")
            contenu = getattr(event, "content", None)

            if contenu is None:
                continue

            for part in getattr(contenu, "parts", []):
                texte = getattr(part, "text", None)

                if not texte or not texte.strip():
                    continue

                texte = texte.strip()

                if texte.upper() in {"SIMPLE", "COMPLEX"}:
                    continue

                if auteur in {
                    "simple_response_agent",
                    "writer_agent",
                }:
                    reponse_finale = texte

        duree = time.perf_counter() - debut

        if not reponse_finale:
            return (
                "⚠️ Aucun agent final n'a fourni de réponse.\n\n"
                f"⏱️ Temps écoulé : {duree:.2f} secondes\n"
                f"⚙️ Mode : {mode}"
            )

        return (
            f"{reponse_finale}\n\n"
            f"---\n"
            f"⏱️ **Temps de génération : {duree:.2f} secondes**\n\n"
            f"⚙️ Mode : **{mode}**"
        )

    except Exception as erreur:
        duree = time.perf_counter() - debut
        traceback.print_exc()

        return (
            "⚠️ **Erreur pendant la génération**\n\n"
            f"**Type :** `{type(erreur).__name__}`\n\n"
            f"**Détail :** `{str(erreur)}`\n\n"
            f"**Temps écoulé :** {duree:.2f} secondes\n\n"
            "Copie ce message pour diagnostiquer le problème."
        )


print("✅ Fonction de diagnostic AYZORA AI prête.")
```

```python
# AYZORA AI — Restauration des composants de base

from google.adk import Workflow, Runner, Context, Event
from google.adk.workflow import START
from google.adk.sessions import InMemorySessionService

# Nom de l'application
APP_NAME = "ayzora_ai"

# Service de sessions
session_service = InMemorySessionService()

print("✅ Workflow importé")
print("✅ Runner importé")
print("✅ START importé")
print("✅ Service de sessions créé")
print("✅ APP_NAME =", APP_NAME)
```

```python
# AYZORA AI — Capture de la demande et routage

from google.adk import Context, Event


def capture_original_request(ctx: Context, node_input: str):
    """Conserve la demande originale pour les agents suivants."""
    ctx.state["original_request"] = node_input
    return Event(output=node_input)


def ai_router_corrige(ctx: Context, node_input: str):
    """Choisit la branche simple ou complexe en conservant la demande."""
    decision = node_input.strip().upper()
    demande_originale = ctx.state.get("original_request", "")

    if decision.startswith("SIMPLE"):
        return Event(output=demande_originale, route="simple")

    if decision.startswith("COMPLEX"):
        return Event(output=demande_originale, route="complex")

    # En cas de décision inattendue, utiliser la branche simple.
    return Event(output=demande_originale, route="simple")


print("✅ Capture de la demande restaurée")
print("✅ Routeur corrigé restauré")
```

```python
# ============================================
# AYZORA AI — Workflow professionnel complexe
# ============================================

professional_routing_workflow = Workflow(
    name="professional_quality_routing_v4",
    edges=[
        (
            START,
            planner_workflow_agent,
            writer_agent,
            quality_control_agent,
        )
    ],
)

print("✅ Workflow complexe restauré !")
print("✅ Planification → Rédaction → Contrôle qualité")
print("✅ Nom :", professional_routing_workflow.name)
```

```python
# AYZORA AI — Reconstruction du workflow principal

ai_routing_workflow_v3 = Workflow(
    name="ai_routing_workflow_v3",
    edges=[
        (
            START,
            capture_original_request,
            orchestrator_agent,
            ai_router_corrige,
            {
                "simple": simple_response_agent,
                "complex": professional_routing_workflow,
            },
        )
    ],
)

# Créer un Runner associé au workflow reconstruit
ai_routing_session_runner = Runner(
    agent=ai_routing_workflow_v3,
    app_name=APP_NAME,
    session_service=session_service,
)

print("✅ Workflow principal reconstruit")
print("✅ Branche simple : simple_response_agent")
print("✅ Branche complexe : professional_quality_routing_v4")
print("✅ Runner restauré")
```

```python
# AYZORA AI — Correction de la récupération des réponses

import uuid
from google.genai import types

async def run_agent(
    demande,
    type_contenu,
    audience,
    langue,
    longueur,
):
    user_id = "ayzora_gradio_user"
    session_id = f"gradio_{uuid.uuid4().hex[:12]}"

    demande_complete = f"""
Demande de l'utilisateur : {demande}
Type de contenu : {type_contenu}
Audience cible : {audience}
Langue : {langue}
Longueur souhaitée : {longueur}
"""

    await session_service.create_session(
        app_name=APP_NAME,
        user_id=user_id,
        session_id=session_id,
        state={
            "revision_count": 0,
            "user_language": langue,
        },
    )

    reponses = []

    async for event in ai_routing_session_runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(
            role="user",
            parts=[types.Part(text=demande_complete)],
        ),
    ):
        if event.content:
            for part in event.content.parts:
                if part.text and part.text.strip():
                    texte = part.text.strip()

                    # Ignorer les décisions internes du routeur.
                    if texte.upper() not in {"SIMPLE", "COMPLEX"}:
                        reponses.append(texte)

    if reponses:
        return "\n\n".join(reponses)

    return (
        "Aucune réponse n'a été récupérée du workflow. "
        "Veuillez consulter les journaux d'exécution."
    )

print("✅ Fonction run_agent corrigée !")
```

```python
# AYZORA AI — Initialisation des composants de base

import os
import uuid
import gradio as gr

from google.adk import Workflow
from google.adk.workflow import START
from google.adk.runners import Runner
from google.adk.sessions import InMemorySessionService
from google.adk.agents import Agent, Context, SequentialAgent
from google.adk.events import Event
from google.adk.models import Gemini
from google.genai import types

# Configuration générale
APP_NAME = "ayzora_ai"
session_service = InMemorySessionService()

# Modèle Gemini avec gestion des tentatives en cas d'erreur
gemini_retry_model = Gemini(
    model="gemini-3.1-flash-lite-preview",
    retry_options=types.HttpRetryOptions(
        attempts=5,
        initial_delay=2.0,
        max_delay=30.0,
        jitter=1.0,
    ),
)

print("✅ Composants de base AYZORA AI initialisés !")
```

```python
# AYZORA AI — Création des cinq agents spécialisés

# 1. ORCHESTRATOR : analyse et oriente les demandes
orchestrator_agent = Agent(
    name="orchestrator_agent",
    model=gemini_retry_model,
    mode="task",
    description="Analyse les demandes et choisit le type de traitement.",
    instruction="""
Tu es l'orchestrateur d'AYZORA AI.

Analyse la demande de l'utilisateur.

Si la demande nécessite une réponse courte, une explication simple
ou une petite publication, réponds exactement : SIMPLE

Si la demande nécessite une stratégie détaillée, un article long,
un plan de contenu ou un travail complexe, réponds exactement : COMPLEX

Ne donne aucune explication. Retourne uniquement SIMPLE ou COMPLEX.
""",
    output_key="request_mode",
)

# 2. PLANNER : prépare le plan
planner_agent = Agent(
    name="planner_agent",
    model=gemini_retry_model,
    mode="task",
    description="Prépare un plan de contenu structuré.",
    instruction="""
Tu es un expert en planification de contenu.

Analyse la demande initiale et prépare un plan adapté.
Pour un contenu long, prévois un titre, une introduction,
4 à 6 sections principales avec leurs idées importantes,
puis une conclusion.

Pour une demande qui ne nécessite pas un article long,
prépare un plan proportionné au besoin.

Réponds toujours en français.
""",
    output_key="content_plan",
)

# 3. VALIDATOR : vérifie le plan
validator_agent = Agent(
    name="validator_agent",
    model=gemini_retry_model,
    mode="task",
    description="Vérifie la pertinence et la structure du plan.",
    instruction="""
Tu es un expert en validation de contenu.

Évalue le plan fourni en fonction de la demande initiale.
Vérifie sa pertinence, sa cohérence, sa structure et l'absence
d'oublis importants.

Si le plan est satisfaisant, commence ta réponse par VALID.
Sinon, commence par RETRY et explique brièvement les corrections
nécessaires.

Réponds toujours en français.
""",
    output_key="validation_result",
)

# 4. WRITER : produit le contenu
writer_agent = Agent(
    name="writer_agent",
    model=gemini_retry_model,
    mode="task",
    description="Rédige le contenu demandé à partir du plan.",
    instruction="""
Tu es un rédacteur professionnel d'AYZORA AI.

DEMANDE INITIALE :
{original_request}

PLAN :
{content_plan}

VALIDATION :
{validation_result?}

RAPPORT DE CONTRÔLE :
{quality_report?}

CONTENU PRÉCÉDENT :
{final_content?}

Rédige le contenu en respectant la demande et le plan.
Si un rapport de contrôle commence par REVISE, corrige les problèmes
signalés en conservant les éléments déjà corrects.
N'invente pas de statistiques ni de faits précis sans source.
Adapte la longueur et le format à la demande de l'utilisateur.
Utilise un français professionnel et naturel.

Retourne uniquement le contenu destiné à l'utilisateur.
""",
    output_key="final_content",
)

# 5. QUALITY CONTROL : vérifie le résultat
quality_control_agent = Agent(
    name="quality_control_agent",
    model=gemini_retry_model,
    mode="task",
    description="Contrôle la qualité du contenu généré.",
    instruction="""
Tu es le responsable du contrôle qualité d'AYZORA AI.

DEMANDE INITIALE :
{original_request?}

PLAN :
{content_plan}

CONTENU :
{final_content}

Vérifie la pertinence, la cohérence, la qualité éditoriale
et le respect de la demande.

Ne prétends pas avoir vérifié des faits sans source fiable.

Si le contenu est satisfaisant, commence par PASS.
Si des corrections importantes sont nécessaires, commence par REVISE
et explique les problèmes de manière concise.

Réponds toujours en français.
""",
    output_key="quality_report",
)

print("✅ Les cinq agents AYZORA AI ont été créés !")

for agent in [
    orchestrator_agent,
    planner_agent,
    validator_agent,
    writer_agent,
    quality_control_agent,
]:
    print("-", agent.name, "| sortie :", agent.output_key)
```

```python
# AYZORA AI — Workflow de planification et validation

content_planning_workflow = SequentialAgent(
    name="content_planning_workflow",
    description="Prépare et valide le plan de contenu.",
    sub_agents=[
        planner_agent,
        validator_agent,
    ],
)

print("✅ Workflow de planification créé !")
print("Agents connectés :")

for agent in content_planning_workflow.sub_agents:
    print("-", agent.name)
```

```python
# AYZORA AI — Workflow professionnel

professional_content_workflow = Workflow(
    name="professional_content_workflow",
    description="Planifie, rédige et contrôle les contenus.",
    edges=[
        (START, content_planning_workflow, writer_agent, quality_control_agent)
    ],
)

print("✅ Workflow professionnel créé !")
print("Ordre des étapes :")
print("1. Planner")
print("2. Validator")
print("3. Writer")
print("4. Quality Control")
```

```python
# AYZORA AI — Runner du workflow professionnel

session_service = InMemorySessionService()

professional_runner = Runner(
    agent=professional_content_workflow,
    app_name=APP_NAME,
    session_service=session_service,
)

print("✅ Runner professionnel créé !")
print("Workflow :", professional_runner.agent.name)
print("Application :", APP_NAME)
```

```python
# AYZORA AI — Orchestration explicite des agents

async def executer_agent(agent, user_id, session_id, message):
    """Exécute un agent et récupère sa réponse finale."""

    await session_service.create_session(
        app_name=APP_NAME,
        user_id=user_id,
        session_id=session_id,
    )

    runner = Runner(
        agent=agent,
        app_name=APP_NAME,
        session_service=session_service,
    )

    reponse = []

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(
            role="user",
            parts=[types.Part(text=message)],
        ),
    ):
        if event.is_final_response() and event.content:
            for part in event.content.parts:
                texte = getattr(part, "text", None)
                if texte:
                    reponse.append(texte)

    return "\n".join(reponse).strip()


async def ayzora_multi_agent(demande):
    """Fait collaborer le Planner, le Validator, le Writer et le QC."""

    uid = f"ayzora_{uuid.uuid4().hex[:8]}"

    print("1/4 — Préparation du plan...")
    plan = await executer_agent(
        planner_agent, uid, f"{uid}_planner",
        f"DEMANDE INITIALE :\n{demande}\n\nProduis un plan détaillé."
    )
    if not plan:
        raise RuntimeError("Le Planner n'a produit aucun résultat.")

    print("2/4 — Validation du plan...")
    validation = await executer_agent(
        validator_agent, uid, f"{uid}_validator",
        f"DEMANDE INITIALE :\n{demande}\n\nPLAN À ÉVALUER :\n{plan}"
    )
    if not validation:
        raise RuntimeError("Le Validator n'a produit aucun résultat.")

    print("3/4 — Rédaction du contenu...")
    contenu = await executer_agent(
        writer_agent, uid, f"{uid}_writer",
        f"DEMANDE INITIALE :\n{demande}\n\n"
        f"PLAN :\n{plan}\n\nVALIDATION :\n{validation}\n\n"
        "Rédige le contenu final en appliquant les corrections nécessaires."
    )
    if not contenu:
        raise RuntimeError("Le Writer n'a produit aucun résultat.")

    print("4/4 — Contrôle qualité...")
    controle = await executer_agent(
        quality_control_agent, uid, f"{uid}_qc",
        f"DEMANDE INITIALE :\n{demande}\n\n"
        f"PLAN :\n{plan}\n\nCONTENU À CONTRÔLER :\n{contenu}"
    )
    if not controle:
        raise RuntimeError("Le Quality Control n'a produit aucun résultat.")

    if controle.strip().upper().startswith("REVISE"):
        print("Révision demandée — nouvelle rédaction...")
        contenu = await executer_agent(
            writer_agent, uid, f"{uid}_writer_revision",
            f"DEMANDE INITIALE :\n{demande}\n\n"
            f"PLAN :\n{plan}\n\n"
            f"PREMIÈRE VERSION :\n{contenu}\n\n"
            f"RAPPORT QUALITÉ :\n{controle}\n\n"
            "Corrige les problèmes signalés et retourne uniquement "
            "la nouvelle version du contenu."
        )

        controle = await executer_agent(
            quality_control_agent, uid, f"{uid}_qc_revision",
            f"DEMANDE INITIALE :\n{demande}\n\n"
            f"CONTENU RÉVISÉ :\n{contenu}"
        )

    return {
        "plan": plan,
        "validation": validation,
        "contenu": contenu,
        "controle_qualite": controle,
    }


print("Orchestration explicite créée.")
```

```python
# AYZORA AI — Test de collaboration des agents

demande_test = (
    "Crée une stratégie marketing de 3 mois pour une agence "
    "immobilière à Dakar. Intègre Facebook, Instagram, LinkedIn "
    "et WhatsApp Business. Propose des actions mensuelles, "
    "des livrables et des indicateurs de performance."
)

try:
    resultat_ayzora = await ayzora_multi_agent(demande_test)

    print("\n" + "=" * 60)
    print("RÉSULTAT FINAL D'AYZORA AI")
    print("=" * 60)

    print("\n--- PLAN ---\n")
    print(resultat_ayzora["plan"])

    print("\n--- VALIDATION ---\n")
    print(resultat_ayzora["validation"])

    print("\n--- CONTENU FINAL ---\n")
    print(resultat_ayzora["contenu"])

    print("\n--- CONTRÔLE QUALITÉ ---\n")
    print(resultat_ayzora["controle_qualite"])

except Exception as erreur:
    print("Erreur pendant le test :")
    print(type(erreur).__name__, str(erreur)[:1500])
```

```python
# AYZORA AI — Correction de la récupération des réponses

async def executer_agent(agent, user_id, session_id, message):
    """Exécute un agent et récupère sa réponse, même si elle n'est pas finale."""

    await session_service.create_session(
        app_name=APP_NAME,
        user_id=user_id,
        session_id=session_id,
    )

    runner = Runner(
        agent=agent,
        app_name=APP_NAME,
        session_service=session_service,
    )

    reponses = []

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(
            role="user",
            parts=[types.Part(text=message)],
        ),
    ):
        print(
            f"Événement : auteur={getattr(event, 'author', None)}, "
            f"final={event.is_final_response()}"
        )

        if event.content and event.content.parts:
            for part in event.content.parts:
                texte = getattr(part, "text", None)
                if texte and texte.strip():
                    reponses.append(texte.strip())

    resultat = "\n".join(reponses).strip()

    if not resultat:
        session = await session_service.get_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
        )
        print("État de session :", session.state)

    return resultat


print("Fonction executer_agent mise à jour.")
```

```python
# ==========================================
# AYZORA AI — Interface professionnelle
# Connectée aux 4 agents spécialisés
# ==========================================

import gradio as gr
import traceback

async def run_ayzora_interface(demande, type_contenu, audience, langue, longueur):
    if not demande or not demande.strip():
        return "Veuillez saisir une demande."

    demande_complete = f"""
DEMANDE DE L'UTILISATEUR :
{demande.strip()}

TYPE DE CONTENU : {type_contenu}
PUBLIC CIBLE : {audience}
LANGUE : {langue}
LONGUEUR SOUHAITÉE : {longueur}

Respecte ces paramètres dans le contenu final.
"""

    try:
        resultat = await ayzora_multi_agent(demande_complete)

        contenu = resultat.get("contenu", "")
        controle = resultat.get("controle_qualite", "")

        if not contenu:
            return "Aucun contenu final n'a été généré."

        return (
            f"# AYZORA AI — Votre contenu\n\n"
            f"{contenu}\n\n"
            f"---\n\n"
            f"### Contrôle qualité\n{controle}"
        )

    except Exception as erreur:
        print(traceback.format_exc())
        return (
            "Une erreur est survenue pendant la génération.\n\n"
            f"Type d'erreur : {type(erreur).__name__}\n"
            f"Détail : {str(erreur)[:500]}"
        )


with gr.Blocks(
    title="AYZORA AI | Studio de contenu intelligent",
    theme=gr.themes.Soft(),
) as professional_demo:

    gr.Markdown(
        """
        # AYZORA AI
        ### Votre studio de création de contenu propulsé par plusieurs agents IA

        Décrivez votre besoin. AYZORA AI planifie, valide, rédige
        et contrôle la qualité du contenu.
        """
    )

    with gr.Row():
        with gr.Column(scale=2):
            demande = gr.Textbox(
                label="Votre demande",
                placeholder=(
                    "Ex. : Crée une stratégie marketing de 3 mois "
                    "pour une agence immobilière à Dakar."
                ),
                lines=5,
            )

            type_contenu = gr.Dropdown(
                choices=[
                    "Automatique",
                    "Publication pour les réseaux sociaux",
                    "Stratégie marketing",
                    "Article de blog",
                    "Calendrier éditorial",
                    "Email professionnel",
                    "Description de produit",
                ],
                value="Automatique",
                label="Type de contenu",
            )

            audience = gr.Textbox(
                label="Public cible",
                placeholder="Ex. : Clients, entreprises, diaspora...",
            )

            with gr.Row():
                langue = gr.Dropdown(
                    choices=["Français", "Anglais", "Wolof"],
                    value="Français",
                    label="Langue",
                )

                longueur = gr.Dropdown(
                    choices=["Court", "Moyen", "Détaillé"],
                    value="Moyen",
                    label="Longueur",
                )

            generer = gr.Button(
                "Générer avec AYZORA AI",
                variant="primary",
            )

        with gr.Column(scale=3):
            resultat = gr.Markdown(
                label="Résultat",
                value="Votre contenu apparaîtra ici.",
            )

    generer.click(
        fn=run_ayzora_interface,
        inputs=[
            demande,
            type_contenu,
            audience,
            langue,
            longueur,
        ],
        outputs=resultat,
        show_progress="full",
    )


print("Interface AYZORA AI créée.")
```

```python
# AYZORA AI — Lancement de l'interface

professional_demo.launch(
    share=True,
    theme=gr.themes.Soft()
)
```

```python
# AYZORA AI — Mode rapide et mode complet

async def run_ayzora_interface(
    demande,
    type_contenu,
    audience,
    langue,
    longueur,
):
    if not demande or not demande.strip():
        return "Veuillez saisir une demande."

    demande_complete = f"""
DEMANDE : {demande.strip()}

TYPE DE CONTENU : {type_contenu}
PUBLIC CIBLE : {audience or "À déterminer selon la demande"}
LANGUE : {langue}
LONGUEUR : {longueur}

Respecte strictement ces paramètres.
Réponds dans la langue demandée.
"""

    type_simple = type_contenu in [
        "Publication pour les réseaux sociaux",
        "Email professionnel",
        "Description de produit",
    ]

    mode_rapide = type_simple or longueur == "Court"

    try:
        if mode_rapide:
            print("⚡ AYZORA AI — Mode rapide")

            uid = f"ayzora_fast_{uuid.uuid4().hex[:8]}"

            contenu = await executer_agent(
                writer_agent,
                uid,
                f"{uid}_writer",
                demande_complete
                + "\nProduis directement le contenu final, sans plan préalable."
            )

            if not contenu:
                return "Aucun contenu généré. Veuillez réessayer."

            return (
                "# ⚡ AYZORA AI — Résultat\n\n"
                + contenu
                + "\n\n---\n*Mode rapide : rédaction directe.*"
            )

        print("AYZORA AI — Mode multi-agents")

        resultat = await ayzora_multi_agent(demande_complete)

        contenu = resultat.get("contenu", "").strip()
        controle = resultat.get("controle_qualite", "").strip()

        if not contenu:
            return "Aucun contenu généré. Veuillez réessayer."

        return (
            "# AYZORA AI — Résultat\n\n"
            + contenu
            + "\n\n---\n\n### Contrôle qualité\n\n"
            + controle
        )

    except Exception as erreur:
        print(type(erreur).__name__, str(erreur)[:1000])
        return (
            f"Une erreur est survenue : {type(erreur).__name__}\n\n"
            f"{str(erreur)[:500]}"
        )


print("Mode rapide et mode multi-agents configurés.")
```

```python
# ==========================================
# AYZORA AI — Moteur d'analyse multimodale
# Images + PDF + Word + Excel + TXT + CSV
# ==========================================

import os
import csv
from pathlib import Path

from pypdf import PdfReader
from docx import Document
from openpyxl import load_workbook
from google.genai import types


def extraire_texte_fichier(chemin):
    """Extrait le texte des formats de documents pris en charge."""

    chemin = Path(chemin)
    extension = chemin.suffix.lower()

    if extension == ".pdf":
        lecteur = PdfReader(str(chemin))
        return "\n".join(
            page.extract_text() or ""
            for page in lecteur.pages
        )

    if extension == ".docx":
        document = Document(str(chemin))
        paragraphes = [
            p.text for p in document.paragraphs if p.text.strip()
        ]
        tableaux = []
        for tableau in document.tables:
            for ligne in tableau.rows:
                tableaux.append(
                    " | ".join(cellule.text for cellule in ligne.cells)
                )
        return "\n".join(paragraphes + tableaux)

    if extension == ".txt":
        return chemin.read_text(encoding="utf-8", errors="replace")

    if extension == ".csv":
        with chemin.open(
            "r", encoding="utf-8-sig", errors="replace", newline=""
        ) as fichier:
            return "\n".join(
                " | ".join(ligne)
                for ligne in csv.reader(fichier)
            )

    if extension == ".xlsx":
        classeur = load_workbook(
            str(chemin),
            read_only=True,
            data_only=True,
        )
        lignes = []
        for feuille in classeur.worksheets:
            lignes.append(f"FEUILLE : {feuille.title}")
            for ligne in feuille.iter_rows(values_only=True):
                valeurs = [str(v) if v is not None else "" for v in ligne]
                if any(valeurs):
                    lignes.append(" | ".join(valeurs))
        classeur.close()
        return "\n".join(lignes)

    return f"[Format non pris en charge pour l'extraction de texte : {extension}]"


def preparer_fichier_multimodal(chemin):
    """Prépare une image ou extrait le texte d'un document."""

    chemin = Path(chemin)
    extension = chemin.suffix.lower()

    extensions_images = {".png", ".jpg", ".jpeg", ".webp"}
    extensions_documents = {".pdf", ".docx", ".txt", ".csv", ".xlsx"}

    if extension in extensions_images:
        image = types.Part.from_bytes(
            data=chemin.read_bytes(),
            mime_type={
                ".png": "image/png",
                ".jpg": "image/jpeg",
                ".jpeg": "image/jpeg",
                ".webp": "image/webp",
            }[extension],
        )
        return [
            types.Part(text=f"Image fournie : {chemin.name}"),
            image,
        ]

    if extension in extensions_documents:
        texte = extraire_texte_fichier(chemin)
        return [
            types.Part(
                text=(
                    f"CONTENU DU DOCUMENT {chemin.name} :\n"
                    f"{texte[:40000]}"
                )
            )
        ]

    return [
        types.Part(
            text=f"Fichier ignoré : format non pris en charge ({chemin.name})."
        )
    ]


async def analyser_fichiers_ayzora(
    demande,
    images_importees=None,
    documents_importes=None,
):
    """Analyse une demande et ses pièces jointes avec Gemini."""

    demande = (demande or "").strip()
    if not demande:
        demande = "Analyse les fichiers fournis et résume les informations importantes."

    parties = [
        types.Part(
            text=(
                "Tu es AYZORA AI, un assistant professionnel. "
                "Réponds en français, sauf si une autre langue est demandée. "
                "Analyse les fichiers réellement fournis. "
                "N'invente pas d'informations absentes des fichiers.\n\n"
                f"DEMANDE :\n{demande}"
            )
        )
    ]

    fichiers = list(images_importees or []) + list(documents_importes or [])

    for fichier in fichiers:
        try:
            parties.extend(preparer_fichier_multimodal(fichier))
        except Exception as erreur:
            parties.append(
                types.Part(
                    text=(
                        f"Impossible de lire {Path(fichier).name} : "
                        f"{type(erreur).__name__}: {str(erreur)[:300]}"
                    )
                )
            )

    reponse = client_multimodal.models.generate_content(
        model="gemini-3.1-flash-lite-preview",
        contents=[types.Content(role="user", parts=parties)],
    )

    return reponse.text or "Gemini n'a retourné aucun texte."


print("Moteur d'analyse multimodale créé.")
print("Formats : PNG, JPG, WEBP, PDF, DOCX, TXT, CSV et XLSX.")
```

```python
# AYZORA AI — Connexion de l'analyse multimodale à Gradio

async def traiter_demande_multimodale(demande, images, documents):
    try:
        resultat = await analyser_fichiers_ayzora(
            demande=demande,
            images_importees=images,
            documents_importes=documents,
        )
        return resultat

    except Exception as erreur:
        import traceback
        traceback.print_exc()

        return (
            f"### Erreur pendant l'analyse\n\n"
            f"**Type :** {type(erreur).__name__}\n\n"
            f"**Détail :** {str(erreur)[:800]}"
        )


with ayzora_multimodal_demo:
    bouton_analyser.click(
        fn=traiter_demande_multimodale,
        inputs=[demande_texte, images, documents],
        outputs=resultat_texte,
        show_progress="full",
    )

print("Bouton d'analyse multimodale connecté.")
```

```python
import asyncio
from google.genai import types

async def tester_ayzora_live():
    async with client_audio.aio.live.connect(
        model="gemini-3.8-live",
        config=config_vocale_ayzora,
    ) as session:

        await session.send_client_content(
            turns=types.Content(
                role="user",
                parts=[
                    types.Part(
                        text="Bonjour AYZORA, présente-toi en une phrase."
                    )
                ],
            ),
            turn_complete=True,
        )

        nombre_blocs_audio = 0
        textes_recus = []

        async for message in session.receive():
            contenu = getattr(message, "server_content", None)

            if contenu:
                transcription = getattr(
                    contenu, "output_transcription", None
                )
                if transcription and transcription.text:
                    textes_recus.append(transcription.text)
                    print("Texte :", transcription.text)

                tour = getattr(contenu, "model_turn", None)
                if tour and tour.parts:
                    for partie in tour.parts:
                        if partie.inline_data and partie.inline_data.data:
                            nombre_blocs_audio += 1

                if getattr(contenu, "turn_complete", False):
                    break

        print("\nNombre de blocs audio reçus :", nombre_blocs_audio)
        print("Transcription disponible :", bool(textes_recus))

await tester_ayzora_live()
```

```python
import numpy as np
import soundfile as sf
import tempfile
from pathlib import Path

def creer_fichier_audio_ayzora(blocs_audio):
    if not blocs_audio:
        return None

    audio_pcm = b"".join(blocs_audio)

    audio_numpy = np.frombuffer(audio_pcm, dtype=np.int16)

    fichier_temporaire = tempfile.NamedTemporaryFile(
        suffix=".wav",
        delete=False
    )
    fichier_temporaire.close()

    sf.write(
        fichier_temporaire.name,
        audio_numpy,
        24000,
        subtype="PCM_16"
    )

    return fichier_temporaire.name

print("Fonction de création audio préparée.")
print("Format de sortie : WAV, PCM 16 bits, 24 000 Hz.")
```

```python
import asyncio
import numpy as np
import soundfile as sf
import tempfile

from google.genai import types


async def conversation_vocale_ayzora(audio_entree):
    if audio_entree is None:
        return "Enregistre d'abord une question.", None

    # Lire l'enregistrement
    audio, frequence = sf.read(audio_entree, dtype="float32")

    # Convertir en mono si nécessaire
    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    # Préparer l'audio PCM 16 bits
    audio_int16 = (np.clip(audio, -1, 1) * 32767).astype(np.int16)

    # Rééchantillonner à 16 kHz pour l'entrée Gemini Live
    if frequence != 16000:
        import scipy.signal
        audio_int16 = scipy.signal.resample_poly(
            audio_int16, 16000, frequence
        ).astype(np.int16)

    texte_recu = []
    blocs_audio = []

    config = types.LiveConnectConfig(
        response_modalities=["AUDIO"],
        system_instruction=(
            "Tu es AYZORA AI, un assistant intelligent et professionnel. "
            "Réponds en français de manière naturelle et utile. "
            "Réponds à la question de l'utilisateur."
        ),
        speech_config=types.SpeechConfig(
            voice_config=types.VoiceConfig(
                prebuilt_voice_config=types.PrebuiltVoiceConfig(
                    voice_name="Kore"
                )
            )
        ),
        output_audio_transcription=types.AudioTranscriptionConfig(),
    )

    async with client_audio.aio.live.connect(
        model="gemini-3.8-live",
        config=config,
    ) as session:

        # Envoyer l'enregistrement par blocs
        taille_bloc = 4096
        audio_bytes = audio_int16.tobytes()

        for debut in range(0, len(audio_bytes), taille_bloc):
            bloc = audio_bytes[debut:debut + taille_bloc]
            await session.send_realtime_input(
                audio=types.Blob(
                    data=bloc,
                    mime_type="audio/pcm;rate=16000",
                )
            )

        # Signaler la fin de l'entrée audio
        await session.send_realtime_input(audio_stream_end=True)

        # Récupérer le texte et la voix
        async for message in session.receive():
            contenu = getattr(message, "server_content", None)

            if contenu:
                transcription = getattr(
                    contenu, "output_transcription", None
                )

                if transcription and transcription.text:
                    texte_recu.append(transcription.text)

                tour = getattr(contenu, "model_turn", None)
                if tour and tour.parts:
                    for partie in tour.parts:
                        donnees = getattr(partie, "inline_data", None)
                        if donnees and donnees.data:
                            blocs_audio.append(donnees.data)

                if getattr(contenu, "turn_complete", False):
                    break

    # Créer le fichier audio de réponse
    fichier_audio = None

    if blocs_audio:
        fichier_audio = creer_fichier_audio_ayzora(blocs_audio)

    texte_final = "".join(texte_recu).strip()

    if not texte_final:
        texte_final = "La réponse audio a été reçue, mais aucune transcription n'est disponible."

    return texte_final, fichier_audio


print("Moteur de conversation vocale AYZORA AI créé.")
```

```python
async def lancer_conversation_vocale(audio):
    try:
        texte, fichier_audio = await conversation_vocale_ayzora(audio)
        return texte, fichier_audio
    except Exception as erreur:
        import traceback
        traceback.print_exc()
        return (
            f"Erreur : {type(erreur).__name__} — {str(erreur)[:500]}",
            None,
        )


with ayzora_multimodal_demo:
    bouton_conversation.click(
        fn=lancer_conversation_vocale,
        inputs=[audio_conversation],
        outputs=[reponse_conversation, audio_reponse],
        show_progress="full",
    )

print("Bouton vocal connecté à AYZORA AI.")
print("Sorties : réponse écrite et réponse audio.")
```

```python
import asyncio
import numpy as np
import soundfile as sf
import tempfile
from google.genai import types


async def conversation_vocale_ayzora(audio_entree, historique=None):
    if audio_entree is None:
        return "Enregistre d'abord une question.", None, historique or []

    historique = list(historique or [])

    # Lire et préparer l'enregistrement
    audio, frequence = sf.read(audio_entree, dtype="float32")

    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    audio_int16 = (np.clip(audio, -1, 1) * 32767).astype(np.int16)

    if frequence != 16000:
        import scipy.signal
        audio_int16 = scipy.signal.resample_poly(
            audio_int16, 16000, frequence
        ).astype(np.int16)

    # Préparer le contexte des échanges précédents
    contexte = "\n".join(
        f"{tour['role']}: {tour['text']}"
        for tour in historique[-12:]
    )

    instruction = (
        "Tu es AYZORA AI, un assistant intelligent et professionnel. "
        "Tu réponds naturellement en français. "
        "Tu dois tenir compte de l'historique fourni pour comprendre "
        "les questions de suivi. Réponds directement à la nouvelle "
        "question, sans répéter inutilement les réponses précédentes.\n\n"
        "HISTORIQUE DE LA CONVERSATION :\n"
        + (contexte if contexte else "C'est le début de la conversation.")
    )

    configuration = types.LiveConnectConfig(
        response_modalities=["AUDIO"],
        system_instruction=instruction,
        speech_config=types.SpeechConfig(
            voice_config=types.VoiceConfig(
                prebuilt_voice_config=types.PrebuiltVoiceConfig(
                    voice_name="Kore"
                )
            )
        ),
        output_audio_transcription=types.AudioTranscriptionConfig(),
    )

    textes = []
    blocs_audio = []

    async with client_audio.aio.live.connect(
        model="gemini-3.8-live",
        config=configuration,
    ) as session:

        # Envoyer la question vocale
        donnees_audio = audio_int16.tobytes()

        for debut in range(0, len(donnees_audio), 4096):
            bloc = donnees_audio[debut:debut + 4096]
            await session.send_realtime_input(
                audio=types.Blob(
                    data=bloc,
                    mime_type="audio/pcm;rate=16000",
                )
            )

        await session.send_realtime_input(audio_stream_end=True)

        # Recevoir la réponse écrite et audio
        async for message in session.receive():
            contenu = getattr(message, "server_content", None)

            if contenu:
                transcription = getattr(
                    contenu, "output_transcription", None
                )

                if transcription and transcription.text:
                    textes.append(transcription.text)

                tour = getattr(contenu, "model_turn", None)

                if tour and tour.parts:
                    for partie in tour.parts:
                        donnees = getattr(partie, "inline_data", None)
                        if donnees and donnees.data:
                            blocs_audio.append(donnees.data)

                if getattr(contenu, "turn_complete", False):
                    break

    reponse = "".join(textes).strip()

    if not reponse:
        reponse = (
            "La réponse audio a été reçue, "
            "mais sa transcription n'est pas disponible."
        )

    # Conserver l'échange pour la prochaine question
    historique.extend([
        {"role": "Utilisateur", "text": "[Question vocale]"},
        {"role": "AYZORA", "text": reponse},
    ])

    fichier_audio = None

    if blocs_audio:
        fichier_audio = creer_fichier_audio_ayzora(blocs_audio)

    return reponse, fichier_audio, historique


print("Moteur vocal avec mémoire conversationnelle préparé.")
```

```python
async def gerer_conversation_vocale(audio, historique):
    historique = list(historique or [])

    if audio is None:
        return (
            historique,
            historique,
            "Enregistre d'abord une question.",
            None,
        )

    try:
        texte, fichier_audio, historique = await conversation_vocale_ayzora(
            audio_entree=audio,
            historique=historique,
        )

        # Ajouter la réponse à l'historique visuel
        historique_affichage = []

        for message in historique:
            if message["role"] == "Utilisateur":
                historique_affichage.append({
                    "role": "user",
                    "content": message["text"],
                })
            else:
                historique_affichage.append({
                    "role": "assistant",
                    "content": message["text"],
                })

        return (
            historique_affichage,
            historique,
            texte,
            fichier_audio,
        )

    except Exception as erreur:
        import traceback
        traceback.print_exc()

        return (
            historique,
            historique,
            f"Erreur : {type(erreur).__name__} — {str(erreur)[:400]}",
            None,
        )


print("Gestionnaire de conversation créé.")
```

```python
import asyncio
import traceback
import numpy as np
import soundfile as sf
import gradio as gr
from google import genai
from google.genai import types


async def repondre_texte_ayzora(message, historique):
    historique = list(historique or [])

    if not message or not message.strip():
        return historique, historique, None, ""

    message = message.strip()

    try:
        # Construire le contexte des derniers échanges
        contexte = "\n".join(
            f"{tour['role']}: {tour['text']}"
            for tour in historique[-12:]
        )

        prompt = f"""
Tu es AYZORA AI, un assistant intelligent, professionnel et serviable.
Tu réponds principalement en français.
Tu réponds précisément à la question actuelle et tu n'inventes pas
d'informations. Utilise l'historique pour comprendre les questions de suivi.

HISTORIQUE :
{contexte if contexte else "Aucun échange précédent."}

NOUVEAU MESSAGE DE L'UTILISATEUR :
{message}

Réponds directement au nouveau message.
"""

        resultat = await asyncio.to_thread(
            client_multimodal.models.generate_content,
            model=MODELE_AYZORA,
            contents=prompt
        )

        reponse = (resultat.text or "").strip()

        if not reponse:
            reponse = "Je n'ai pas réussi à produire une réponse. Réessaie."

        historique.extend([
            {"role": "Utilisateur", "text": message},
            {"role": "AYZORA", "text": reponse}
        ])

        affichage = [
            {
                "role": "user" if tour["role"] == "Utilisateur" else "assistant",
                "content": tour["text"]
            }
            for tour in historique
        ]

        return affichage, historique, None, ""

    except Exception as erreur:
        traceback.print_exc()
        return (
            [
                {
                    "role": "user" if tour["role"] == "Utilisateur" else "assistant",
                    "content": tour["text"]
                }
                for tour in historique
            ],
            historique,
            None,
            f"Erreur Gemini : {type(erreur).__name__} — {str(erreur)[:250]}"
        )


print("✅ Moteur de conversation écrite préparé.")
```

```python
import asyncio
import traceback

MODELE_AYZORA = "models/gemini-3.8-flash"

async def repondre_texte_ayzora(message, historique):
    historique = list(historique or [])

    if not message or not message.strip():
        affichage = [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in historique
        ]
        return affichage, historique, None, ""

    message = message.strip()

    contexte = "\n".join(
        f"{t['role']}: {t['text']}"
        for t in historique[-12:]
    )

    prompt = f"""
Tu es AYZORA AI, un assistant intelligent et professionnel.
Tu réponds en français et tu réponds précisément à la question posée.
Tu utilises l'historique pour comprendre les questions de suivi.

HISTORIQUE :
{contexte if contexte else "Début de conversation."}

QUESTION ACTUELLE :
{message}

Réponds directement à la question actuelle.
"""

    try:
        resultat = await asyncio.to_thread(
            client_multimodal.models.generate_content,
            model=MODELE_AYZORA,
            contents=prompt
        )

        reponse = (resultat.text or "").strip()

        if not reponse:
            reponse = "Je n'ai pas pu générer une réponse. Réessaie."

        historique.extend([
            {"role": "Utilisateur", "text": message},
            {"role": "AYZORA", "text": reponse}
        ])

        affichage = [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in historique
        ]

        return affichage, historique, None, ""

    except Exception as erreur:
        traceback.print_exc()
        return (
            [
                {
                    "role": "user" if t["role"] == "Utilisateur" else "assistant",
                    "content": t["text"]
                }
                for t in historique
            ],
            historique,
            None,
            f"Erreur : {type(erreur).__name__} — {str(erreur)[:300]}"
        )


# Mettre à jour les callbacks écrits déjà enregistrés dans Gradio
nb = 0
for entree in ayzora_chat_demo.fns.values():
    fonction = getattr(entree, "fn", None)
    if getattr(fonction, "__name__", "") == "repondre_texte_ayzora":
        entree.fn = repondre_texte_ayzora
        nb += 1

print("✅ Fonction corrigée avec", MODELE_AYZORA)
print("Callbacks Gradio mis à jour :", nb)
```

```python
import asyncio
import traceback

MODELE_AYZORA = "models/gemini-3.8-flash"
MODELE_SECOURS_AYZORA = "models/gemini-3.1-flash-lite-preview"

async def repondre_texte_ayzora(message, historique):
    historique = list(historique or [])

    def affichage_chat(tours):
        return [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in tours
        ]

    if not message or not message.strip():
        return affichage_chat(historique), historique, None, ""

    message = message.strip()

    contexte = "\n".join(
        f"{t['role']}: {t['text']}"
        for t in historique[-12:]
    )

    prompt = f"""
Tu es AYZORA AI, un assistant intelligent et professionnel.
Réponds en français, directement et précisément.
Utilise l'historique pour comprendre les questions de suivi.
N'invente pas de faits.

HISTORIQUE :
{contexte if contexte else "Début de conversation."}

QUESTION ACTUELLE :
{message}
"""

    erreur_finale = None

    for modele in [MODELE_AYZORA, MODELE_SECOURS_AYZORA]:
        try:
            resultat = await asyncio.wait_for(
                asyncio.to_thread(
                    client_multimodal.models.generate_content,
                    model=modele,
                    contents=prompt
                ),
                timeout=30
            )

            reponse = (resultat.text or "").strip()
            if not reponse:
                raise ValueError("Le modèle a renvoyé une réponse vide.")

            historique.extend([
                {"role": "Utilisateur", "text": message},
                {"role": "AYZORA", "text": reponse}
            ])

            return affichage_chat(historique), historique, None, ""

        except Exception as erreur:
            erreur_finale = erreur
            print(f"Modèle {modele} indisponible : {type(erreur).__name__}")

    message_erreur = (
        "AYZORA rencontre momentanément une indisponibilité des modèles IA. "
        "Ta question n'a pas été ajoutée à l'historique. Réessaie dans quelques instants."
    )

    print("Détail technique :", repr(erreur_finale))

    return affichage_chat(historique), historique, None, message_erreur


print("✅ Moteur AYZORA mis à jour : modèle principal, modèle de secours et délai limité.")
```

```python
import asyncio
import traceback

MODELE_AYZORA = "models/gemini-3.1-flash-lite-preview"

async def repondre_texte_ayzora(message, historique):
    historique = list(historique or [])

    def formater_chat(tours):
        return [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in tours
        ]

    if not message or not message.strip():
        return formater_chat(historique), historique, None, ""

    message = message.strip()

    contexte = "\n".join(
        f"{t['role']}: {t['text']}"
        for t in historique[-12:]
    )

    prompt = f"""
Tu es AYZORA AI, un assistant intelligent et professionnel.
Tu réponds en français, de manière claire et pertinente.
Tu tiens compte de l'historique pour comprendre les questions de suivi.
Tu ne prétends pas avoir consulté des annonces immobilières si tu ne l'as pas fait.

HISTORIQUE :
{contexte if contexte else "Début de conversation."}

QUESTION ACTUELLE :
{message}

Réponds directement à la question actuelle.
"""

    try:
        resultat = await asyncio.wait_for(
            asyncio.to_thread(
                client_multimodal.models.generate_content,
                model=MODELE_AYZORA,
                contents=prompt
            ),
            timeout=30
        )

        reponse = (resultat.text or "").strip()

        if not reponse:
            reponse = "Je n'ai pas reçu de réponse exploitable. Réessaie."

        historique.extend([
            {"role": "Utilisateur", "text": message},
            {"role": "AYZORA", "text": reponse}
        ])

        return formater_chat(historique), historique, None, ""

    except Exception as erreur:
        traceback.print_exc()
        return (
            formater_chat(historique),
            historique,
            None,
            f"AYZORA est momentanément indisponible. Réessaie dans quelques instants. Détail : {type(erreur).__name__}"
        )


# Mettre à jour les deux événements écrits déjà enregistrés dans Gradio
nb = 0

for entree in ayzora_chat_demo.fns.values():
    fonction = getattr(entree, "fn", None)

    if getattr(fonction, "__name__", "") == "repondre_texte_ayzora":
        entree.fn = repondre_texte_ayzora
        nb += 1

print("✅ Modèle écrit stabilisé :", MODELE_AYZORA)
print("Événements mis à jour :", nb)
```

```python
import asyncio
import traceback

async def traiter_message_unique(message, fichiers, historique):
    historique = list(historique or [])

    def formater_chat(tours):
        return [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in tours
        ]

    if not (message and message.strip()) and not fichiers:
        return formater_chat(historique), historique, "", None, "Écris un message ou joins un fichier."

    try:
        contenu = (message or "").strip()

        # Analyser les fichiers joints avec la fonction existante
        if fichiers:
            liste_fichiers = fichiers if isinstance(fichiers, list) else [fichiers]

            resultat_fichiers = await asyncio.to_thread(
                analyser_fichiers_ayzora,
                contenu or "Analyse ces fichiers et explique leur contenu.",
                [],
                liste_fichiers
            )

            # Récupérer le résultat sans supposer son format
            if isinstance(resultat_fichiers, (tuple, list)):
                analyse = str(resultat_fichiers[0])
            else:
                analyse = str(resultat_fichiers)

            contenu = (
                f"{contenu}\n\nAnalyse des fichiers joints :\n{analyse}"
                if contenu else analyse
            )

        if not contenu:
            contenu = "Analyse les fichiers joints."

        affichage, historique, _, erreur = await repondre_texte_ayzora(
            contenu, historique
        )

        return affichage, historique, "", None, erreur or ""

    except Exception as erreur:
        traceback.print_exc()
        return (
            formater_chat(historique),
            historique,
            message or "",
            None,
            f"Erreur : {type(erreur).__name__} — {str(erreur)[:250]}"
        )


print("✅ Traitement des messages et fichiers préparé.")
```

```python
import asyncio
import traceback

async def traiter_message_unique(message, fichiers, historique):
    historique = list(historique or [])

    def formater_chat(tours):
        return [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in tours
        ]

    if not (message and message.strip()) and not fichiers:
        return (
            formater_chat(historique),
            historique,
            "",
            None,
            "Écris un message ou joins un fichier."
        )

    try:
        contenu = (message or "").strip()

        if fichiers:
            liste_fichiers = (
                fichiers if isinstance(fichiers, list) else [fichiers]
            )

            images = []
            documents = []

            for fichier in liste_fichiers:
                nom = str(fichier).lower()

                if nom.endswith((".png", ".jpg", ".jpeg", ".webp")):
                    images.append(fichier)
                else:
                    documents.append(fichier)

            analyse = await analyser_fichiers_ayzora(
                demande=contenu or "Analyse les fichiers joints.",
                images_importees=images,
                documents_importes=documents
            )

            contenu = (
                f"{contenu}\n\nAnalyse des fichiers :\n{analyse}"
                if contenu else analyse
            )

        affichage, historique, _, erreur = await repondre_texte_ayzora(
            contenu,
            historique
        )

        return affichage, historique, "", None, erreur or ""

    except Exception as erreur:
        traceback.print_exc()

        return (
            formater_chat(historique),
            historique,
            message or "",
            None,
            f"Erreur : {type(erreur).__name__} — {str(erreur)[:250]}"
        )


print("✅ Traitement des messages et fichiers corrigé.")
```

```python
with ayzora_interface_finale:
    bouton_envoi_unique.click(
        fn=traiter_message_unique,
        inputs=[
            texte_unique,
            fichiers_uniques,
            memoire_unique
        ],
        outputs=[
            chat_unique,
            memoire_unique,
            texte_unique,
            audio_sortie_unique,
            gr.Textbox(visible=False)
        ]
    )

    texte_unique.submit(
        fn=traiter_message_unique,
        inputs=[
            texte_unique,
            fichiers_uniques,
            memoire_unique
        ],
        outputs=[
            chat_unique,
            memoire_unique,
            texte_unique,
            audio_sortie_unique,
            gr.Textbox(visible=False)
        ]
    )

print("✅ Envoi des messages et fichiers connecté.")
```

```python
with ayzora_interface_finale:

    # Afficher le microphone dans la même interface
    bouton_micro_unique.click(
        fn=lambda: gr.update(visible=True),
        inputs=[],
        outputs=[audio_unique]
    )

    # Envoyer la question vocale dans la conversation commune
    audio_unique.change(
        fn=repondre_vocal_ayzora,
        inputs=[
            audio_unique,
            memoire_unique
        ],
        outputs=[
            chat_unique,
            memoire_unique,
            audio_sortie_unique
        ]
    )

    # Réinitialiser toute la conversation
    bouton_nouvelle_unique.click(
        fn=lambda: (
            [],
            [],
            "",
            None,
            None,
            gr.update(visible=False)
        ),
        inputs=[],
        outputs=[
            chat_unique,
            memoire_unique,
            texte_unique,
            fichiers_uniques,
            audio_sortie_unique,
            audio_unique
        ]
    )

print("✅ Microphone et nouvelle conversation connectés.")
```

```python
import requests
from google.colab import userdata

SUPABASE_URL = userdata.get("SUPABASE_URL").rstrip("/")
SUPABASE_KEY = userdata.get("SUPABASE_PUBLISHABLE_KEY")

def charger_connaissances_supabase(entreprise=None):
    url = f"{SUPABASE_URL}/rest/v1/ayzora_knowledge"

    params = {
        "select": "company,category,title,content,verified,updated_at",
        "order": "updated_at.desc"
    }

    if entreprise:
        params["company"] = f"eq.{entreprise}"

    headers = {
        "apikey": SUPABASE_KEY
    }

    response = requests.get(
        url,
        headers=headers,
        params=params,
        timeout=30
    )

    response.raise_for_status()
    return response.json()

print("Fonction de récupération créée avec succès.")
```

```python
import json

def construire_contexte_supabase(question):
    """Récupère les connaissances persistantes pour AYZORA AI."""
    donnees = charger_connaissances_supabase()

    contexte = []

    for element in donnees:
        contexte.append({
            "entreprise": element["company"],
            "titre": element["title"],
            "connaissances": element["content"]
        })

    return json.dumps(
        contexte,
        ensure_ascii=False,
        indent=2
    )

print("✅ Fonction de contexte Supabase créée.")
```

```python
async def repondre_texte_ayzora(message, historique):
    import asyncio
    import json
    import traceback

    try:
        # Récupérer les connaissances depuis Supabase
        donnees = await asyncio.to_thread(
            charger_connaissances_supabase
        )

        contexte_entreprises = json.dumps(
            [
                {
                    "entreprise": element["company"],
                    "titre": element["title"],
                    "connaissances": element["content"]
                }
                for element in donnees
            ],
            ensure_ascii=False,
            indent=2
        )

        # Préparer l'historique de la conversation
        contexte = "\n".join(
            f"{element['role']} : {element['text']}"
            for element in historique
        )

        prompt = f"""
{INSTRUCTIONS_AYZORA}

BASE DE CONNAISSANCES RÉCUPÉRÉE DE SUPABASE :
{contexte_entreprises}

RÈGLES IMPORTANTES :
- Utilise les connaissances récupérées depuis Supabase.
- Une valeur null signifie que l'information n'est pas renseignée.
- Ne transforme jamais une information manquante en fait.
- Pour une question générale, réponds normalement.
- Pour une question sur une entreprise, utilise sa propre fiche.
- Si tu ne connais pas la réponse, dis-le clairement.

HISTORIQUE DE LA CONVERSATION :
{contexte if contexte else "Début de conversation."}

MESSAGE ACTUEL :
{message}

Réponds directement au message actuel.
"""

        resultat = await asyncio.wait_for(
            asyncio.to_thread(
                client_multimodal.models.generate_content,
                model="models/gemini-3.1-flash-lite-preview",
                contents=prompt
            ),
            timeout=30
        )

        reponse = (resultat.text or "").strip()

        if not reponse:
            reponse = "Je n'ai pas obtenu de réponse exploitable. Réessaie."

        historique.extend([
            {"role": "Utilisateur", "text": message},
            {"role": "AYZORA", "text": reponse}
        ])

        return formater_chat(historique), historique, None, ""

    except Exception as erreur:
        traceback.print_exc()

        return (
            formater_chat(historique),
            historique,
            None,
            f"AYZORA est momentanément indisponible : {type(erreur).__name__}."
        )
```

```python
import socket

def trouver_port_libre(debut=7870, fin=7900):
    for port in range(debut, fin + 1):
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
            if sock.connect_ex(("127.0.0.1", port)) != 0:
                return port
    raise RuntimeError("Aucun port libre trouvé.")

PORT_AYZORA = trouver_port_libre()

print("✅ Port libre trouvé :", PORT_AYZORA)
```

```python
def formater_chat(historique):
    """Formate l'historique pour l'affichage dans Gradio."""
    messages = []

    for element in historique:
        role = element.get("role", "")
        texte = element.get("text", "")

        if role == "Utilisateur":
            messages.append({
                "role": "user",
                "content": texte
            })
        elif role == "AYZORA":
            messages.append({
                "role": "assistant",
                "content": texte
            })

    return messages

print("✅ Fonction formater_chat créée.")
```

```python
async def repondre_vocal_ayzora(audio_entree, historique):
    historique = list(historique or [])

    if audio_entree is None:
        affichage = [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in historique
        ]
        return affichage, historique, None

    try:
        # Transcrire la question vocale
        resultat_transcription = await asyncio.to_thread(
            client_multimodal.models.generate_content,
            model="gemini-2.5-flash",
            contents=[
                types.Part.from_bytes(
                    data=open(audio_entree, "rb").read(),
                    mime_type="audio/wav"
                ),
                "Transcris fidèlement cette question audio en français. "
                "Retourne uniquement les mots prononcés."
            ]
        )

        question = (resultat_transcription.text or "").strip()

        if not question:
            raise ValueError("La question vocale n'a pas pu être transcrite.")

        # Répondre avec le même historique que le mode écrit
        affichage, historique, _, erreur = await repondre_texte_ayzora(
            question, historique
        )

        if erreur:
            raise RuntimeError(erreur)

        # Récupérer la dernière réponse
        reponse = historique[-1]["text"]

        # Générer une réponse vocale avec Gemini TTS
        resultat_audio = await asyncio.to_thread(
            client_multimodal.models.generate_content,
            model="gemini-2.5-flash-preview-tts",
            contents=f"""
Lis naturellement en français, avec une voix claire et professionnelle,
la réponse suivante d'AYZORA AI :

{reponse}
""",
            config=types.GenerateContentConfig(
                response_modalities=["AUDIO"],
                speech_config=types.SpeechConfig(
                    voice_config=types.VoiceConfig(
                        prebuilt_voice_config=types.PrebuiltVoiceConfig(
                            voice_name="Kore"
                        )
                    )
                )
            )
        )

        blocs_audio = []
        for candidat in (resultat_audio.candidates or []):
            for partie in (candidat.content.parts or []):
                if partie.inline_data and partie.inline_data.data:
                    blocs_audio.append(partie.inline_data.data)

        fichier_audio = None

        if blocs_audio:
            audio_pcm = b"".join(blocs_audio)
            audio_numpy = np.frombuffer(audio_pcm, dtype=np.int16)

            fichier_temp = tempfile.NamedTemporaryFile(
                suffix=".wav", delete=False
            )
            fichier_temp.close()

            sf.write(
                fichier_temp.name,
                audio_numpy,
                24000,
                subtype="PCM_16"
            )
            fichier_audio = fichier_temp.name

        return affichage, historique, fichier_audio

    except Exception as erreur:
        traceback.print_exc()

        affichage = [
            {
                "role": "user" if t["role"] == "Utilisateur" else "assistant",
                "content": t["text"]
            }
            for t in historique
        ]

        return affichage, historique, None


print("✅ Fonction vocale préparée.")
```

```python
from google.adk import Context, Event

# 1. Conserver la demande originale de l'utilisateur
def capture_original_request(ctx: Context, node_input: str):
    ctx.state["original_request"] = node_input
    return Event(output=node_input)


# 2. Router : choisir la branche appropriée
def ai_router(ctx: Context, node_input: str):

    decision = node_input.strip().upper()

    # Récupérer la demande originale
    original_request = ctx.state.get(
        "original_request",
        node_input
    )

    if decision.startswith("SIMPLE"):
        return Event(
            output=original_request,
            route="simple"
        )

    if decision.startswith("COMPLEX"):
        return Event(
            output=original_request,
            route="complex"
        )

    # Par défaut, traiter comme une demande simple
    return Event(
        output=original_request,
        route="simple"
    )


print("📦 Capture de la demande originale restaurée !")
print("🔀 AI Router restauré !")
```

```python
import gradio as gr
from google.genai import types
from datetime import datetime
from zoneinfo import ZoneInfo


async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
    progress=gr.Progress(),
    request: gr.Request = None,
):
    """Traite la demande avec le AI Routing Workflow V3."""

    now = datetime.now(ZoneInfo("Africa/Dakar"))
    current_date = now.strftime("%d/%m/%Y")
    current_day = now.strftime("%A")

    progress(0.05, desc="🧠 Analyse de votre demande...")
    session_hash = request.session_hash if request else "default"

    if session_hash not in gradio_sessions:
        user_id = f"gradio_user_{session_hash}"
        session_id = f"gradio_session_{session_hash}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        gradio_sessions[session_hash] = {
            "user_id": user_id,
            "session_id": session_id,
        }

    user_id = gradio_sessions[session_hash]["user_id"]
    session_id = gradio_sessions[session_hash]["session_id"]

    prompt = f"""
DATE ACTUELLE : {current_date}
JOUR ACTUEL : {current_day}
FUSEAU HORAIRE : Africa/Dakar

DEMANDE DE L'UTILISATEUR :
{user_message}

TYPE DE CONTENU :
{content_type}

AUDIENCE CIBLE :
{audience}

LANGUE DEMANDÉE :
{language}

LONGUEUR :
{length}

Consignes :
- Réponds précisément à la demande.
- Respecte le type de contenu, l'audience et la longueur.
- Rédige entièrement dans la langue sélectionnée.
- Pour le français, utilise le français.
- Pour English, utilise l'anglais.
- Pour Wolof, utilise un wolof naturel et compréhensible.
- Utilise la date actuelle fournie ci-dessus pour les références temporelles.
- Distingue les faits confirmés, les prévisions et les informations incertaines.
- N'invente jamais de dates, de statistiques, de prix ou de faits.
- Pour les informations récentes, vérifie les sources avant de les présenter comme exactes.
- Si une information ne peut pas être vérifiée, indique-le clairement.
- Ne mentionne pas le fonctionnement interne des agents.
- Retourne uniquement le contenu destiné à l'utilisateur.
"""

    message = types.Content(
        role="user",
        parts=[types.Part(text=prompt)],
    )

    final_response = None

    async for event in ai_routing_session_runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=message,
    ):
        author = getattr(event, "author", "")

        if author == "orchestrator_agent":
            progress(0.15, desc="🧠 Analyse de la demande...")
        elif author == "simple_response_agent":
            progress(0.65, desc="⚡ Génération de la réponse...")
        elif author == "planner_workflow_agent":
            progress(0.35, desc="📋 Planification...")
        elif author == "writer_agent":
            progress(0.60, desc="✍️ Rédaction...")
        elif author == "quality_control_agent":
            progress(0.80, desc="🛡️ Contrôle qualité...")

        if author in ("simple_response_agent", "writer_agent"):
            content = getattr(event, "content", None)

            if content:
                for part in getattr(content, "parts", []):
                    if getattr(part, "text", None):
                        final_response = part.text

    progress(1.0, desc="✅ Traitement terminé !")

    return final_response or "Aucune réponse n'a été générée."


print("🤖 run_agent restauré et connecté au Workflow V3 !")
```

```python
import gradio as gr
from google.genai import types
from datetime import datetime
from zoneinfo import ZoneInfo

async def run_agent(
    user_message,
    content_type,
    audience,
    language,
    length,
    progress=gr.Progress(),
    request: gr.Request = None,
):
    """Traite la demande avec le AI Routing Workflow V3."""

    now = datetime.now(ZoneInfo("Africa/Dakar"))
    current_date = now.strftime("%d/%m/%Y")
    current_day = now.strftime("%A")

    progress(0.05, desc="🧠 Analyse de votre demande...")
    session_hash = request.session_hash if request else "default"

    if session_hash not in gradio_sessions:
        user_id = f"gradio_user_{session_hash}"
        session_id = f"gradio_session_{session_hash}"

        await session_service.create_session(
            app_name=APP_NAME,
            user_id=user_id,
            session_id=session_id,
            state={
                "revision_count": 0,
                "user_language": language,
            },
        )

        gradio_sessions[session_hash] = {
            "user_id": user_id,
            "session_id": session_id,
        }

    user_id = gradio_sessions[session_hash]["user_id"]
    session_id = gradio_sessions[session_hash]["session_id"]

    prompt = f"""
DATE ACTUELLE : {current_date}
JOUR ACTUEL : {current_day}
FUSEAU HORAIRE : Africa/Dakar

DEMANDE DE L'UTILISATEUR :
{user_message}

TYPE DE CONTENU :
{content_type}

AUDIENCE CIBLE :
{audience}

LANGUE DEMANDÉE :
{language}

LONGUEUR :
{length}

Consignes :
- Réponds précisément à la demande.
- Respecte le type de contenu, l'audience et la longueur.
- Rédige entièrement dans la langue sélectionnée.
- Pour le français, utilise le français.
- Pour English, utilise l'anglais.
- Pour Wolof, utilise un wolof naturel et compréhensible.
- Utilise la date actuelle fournie ci-dessus pour les références temporelles.
- Distingue les faits confirmés, les prévisions et les informations incertaines.
- N'invente jamais de dates, de statistiques, de prix ou de faits.
- Pour les informations récentes, vérifie les sources avant de les présenter comme exactes.
- Si une information ne peut pas être vérifiée, indique-le clairement.
- Ne mentionne pas le fonctionnement interne des agents.
- Retourne uniquement le contenu destiné à l'utilisateur.
"""

    message = types.Content(
        role="user",
        parts=[types.Part(text=prompt)],
    )

    final_response = None

    try:
        async for event in ai_routing_session_runner.run_async(
            user_id=user_id,
            session_id=session_id,
            new_message=message,
        ):
            author = getattr(event, "author", "")

            if author == "orchestrator_agent":
                progress(0.15, desc="🧠 Analyse de la demande...")
            elif author == "simple_response_agent":
                progress(0.65, desc="⚡ Génération de la réponse...")
            elif author == "planner_workflow_agent":
                progress(0.35, desc="📋 Planification...")
            elif author == "writer_agent":
                progress(0.60, desc="✍️ Rédaction...")
            elif author == "quality_control_agent":
                progress(0.80, desc="🛡️ Contrôle qualité...")

            if author in ("simple_response_agent", "writer_agent"):
                content = getattr(event, "content", None)

                if content:
                    for part in getattr(content, "parts", []):
                        texte = getattr(part, "text", None)
                        if texte:
                            final_response = texte

        progress(1.0, desc="✅ Traitement terminé !")

        return final_response or (
            "Aucune réponse n'a été générée. "
            "Vérifie les journaux du workflow pour plus de détails."
        )

    except Exception as erreur:
        print(f"Erreur dans run_agent : {type(erreur).__name__}: {erreur}")
        return (
            "Une erreur est survenue pendant le traitement. "
            "Consulte les journaux Colab pour identifier le problème."
        )

print("✅ Fonction run_agent restaurée.")
```

```python
# Afficher uniquement le texte final produit par le workflow

final_event = None

for event in reversed(result):
    if getattr(event, "content", None):
        parts = getattr(event.content, "parts", [])
        for part in parts:
            if getattr(part, "text", None):
                final_event = part.text
                break
    if final_event:
        break

if final_event:
    print("=" * 70)
    print("📄 CONTENU FINAL")
    print("=" * 70)
    print(final_event)
else:
    print("⚠️ Aucun contenu texte final trouvé.")
```

```python
for i, event in enumerate(result):
    author = getattr(event, "author", "inconnu")

    text_content = ""

    if getattr(event, "content", None):
        for part in getattr(event.content, "parts", []):
            if getattr(part, "text", None):
                text_content += part.text

    print(f"\n--- EVENT {i} | AUTHOR: {author} ---")
    print(text_content[:300])
```

```python
# Afficher uniquement le contenu produit par le Writer Agent

writer_content = None

for event in result:
    if getattr(event, "author", "") == "writer_agent":
        if getattr(event, "content", None):
            parts = getattr(event.content, "parts", [])

            for part in parts:
                if getattr(part, "text", None):
                    writer_content = part.text
                    break

    if writer_content:
        break

if writer_content:
    print("=" * 70)
    print("📄 ARTICLE FINAL")
    print("=" * 70)
    print(writer_content)
else:
    print("⚠️ Aucun contenu du Writer Agent trouvé.")
```

```python
from google.adk import Event
from google.adk.events import EventActions

def revision_counter(node_input: str):
    """
    Initialise ou incrémente le compteur de révisions.
    """

    return Event(
        output=node_input,
        actions=EventActions(
            state_delta={
                "revision_count": 1
            }
        )
    )

print("🔢 Revision Counter Node créé !")
```

```python
quality_runner = InMemoryRunner(
    agent=professional_content_quality_workflow
)

quality_workflow_result = await quality_runner.run_debug(
    """
    Crée un article professionnel sur le thème :

    Comment les entreprises africaines peuvent utiliser
    les AI Agents pour améliorer leur productivité,
    leur service client et leur croissance.

    Le contenu doit être destiné à des dirigeants
    et responsables d'entreprises.
    """
)

print(quality_workflow_result)
```

```python
from google.adk import Context, Event

def quality_router(ctx: Context, node_input: str):
    """
    Route le workflow selon le Quality Control
    et limite le nombre de révisions.
    """

    quality_report = node_input.strip()

    # ─────────────────────────────
    # CAS 1 : contenu validé
    # ─────────────────────────────
    if quality_report.upper().startswith("PASS"):
        return Event(
            output=quality_report,
            route="pass"
        )

    # ─────────────────────────────
    # CAS 2 : demande de révision
    # ─────────────────────────────
    current_count = ctx.state.get("revision_count", 0)

    # Limite atteinte
    if current_count >= MAX_REVISIONS:
        return Event(
            output=quality_report,
            route="max_revisions"
        )

    # Incrémentation du compteur
    ctx.state["revision_count"] = current_count + 1

    return Event(
        output=quality_report,
        route="revise"
    )


print("🔄 Quality Router avec limite de révisions configuré !")
```

```python
from google.adk import Event

def max_revisions_node(node_input: str):
    """
    Termine le workflow lorsque la limite de révisions est atteinte.
    """
    return Event(
        output=(
            "MAX_REVISIONS\n\n"
            "Le contenu n'a pas obtenu la validation finale "
            "après le nombre maximal de révisions autorisées.\n\n"
            f"Rapport final du Quality Control :\n{node_input}"
        )
    )

print("🛑 Max Revisions Node créé !")
```

```python
test_pass = quality_router(
    "PASS\nLe contenu est suffisamment professionnel."
)

print("Route :", test_pass.actions.route)
print("Output :", test_pass.output)
```

```python
test_revise = quality_router(
    "REVISE\nLa conclusion doit être améliorée et certaines sections manquent de profondeur."
)

print("Route :", test_revise.actions.route)
print("Output :", test_revise.output)
```

```python
from google.adk import Workflow
from google.adk.workflow import START
from google.adk.events import Event


def final_pass_node(node_input):
    """
    Termine le workflow lorsque le Quality Control valide le contenu.
    """
    return Event(
        output=node_input
    )


test_routing_workflow = Workflow(
    name="test_quality_routing",
    edges=[
        (
            START,
            quality_router,
            {
                "pass": final_pass_node,
                "revise": writer_agent
            }
        )
    ],
)

print("🔀 Workflow de routage PASS / REVISE créé !")
```

```python
routing_runner = InMemoryRunner(
    agent=test_routing_workflow
)

routing_result = await routing_runner.run_debug(
    "PASS\nLe contenu est suffisamment professionnel."
)

print(routing_result)
```
